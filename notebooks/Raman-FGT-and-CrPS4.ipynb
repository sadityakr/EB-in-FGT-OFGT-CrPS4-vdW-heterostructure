```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path().resolve().parent))

from scripts.notebook_setup import setup
PROJECT_ROOT, np, pd, plt, Path = setup()
```

```python
RAMAN_DIR = PROJECT_ROOT / "data" / "RAMAN"
X_MIN = 80  # cm⁻¹ — crops Rayleigh scattering leaking through the notch filter
```

```python
d = np.loadtxt(RAMAN_DIR / "sample1.txt")
x, y = d[d[:, 0] >= X_MIN, 0], d[d[:, 0] >= X_MIN, 1]

fig, ax = plt.subplots(figsize=(6, 5))
ax.plot(x, y, color="#E69F00", lw=0.9)
ax.set_xlabel(r"Raman shift (cm$^{-1}$)")
ax.set_ylabel("Intensity (counts)")
ax.set_title("Fe$_3$GeTe$_2$ bulk crystal")
ax.set_xlim(X_MIN, 500)
fig.tight_layout()
plt.show()
```

```python
d = np.loadtxt(RAMAN_DIR / "sample2.txt")
x, y = d[d[:, 0] >= X_MIN, 0], d[d[:, 0] >= X_MIN, 1]

fig, ax = plt.subplots(figsize=(6, 5))
ax.plot(x, y, color="#56B4E9", lw=0.9)
ax.set_xlabel(r"Raman shift (cm$^{-1}$)")
ax.set_ylabel("Intensity (counts)")
ax.set_title("CrPS$_4$ bulk crystal")
ax.set_xlim(X_MIN, 700)
fig.tight_layout()
plt.show()
```
