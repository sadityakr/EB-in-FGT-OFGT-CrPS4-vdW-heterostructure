# $V_{xy}$ and $V_{xx}$ vs $T$ — FGT/O (ACS Nano rebuttal data)

**Device:** FGT/O-FGT


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path().resolve().parent))

from scripts.notebook_setup import setup
PROJECT_ROOT, np, pd, plt, Path = setup()


```

```python

DATA_FILE = PROJECT_ROOT / "data" / "Rebuttal_FGT_CPS" / "FGT-O_RvsT_FCW_0.1T_Vxy_Vxx.txt"

df = pd.read_csv(DATA_FILE, sep="\t", comment="#", header=0)
# columns: I_A, T_K, Vxy (V), V_xx (V)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(df["T_K"], df["Vxy (V)"] * 1e5, lw=1.5)
axes[0].set_xlabel(r"$T$ (K)")
axes[0].set_ylabel(r"$R_{xy}$ (Ohm)")
axes[0].set_title("FGT/OFGT — Hall voltage")

axes[1].plot(df["T_K"], df["V_xx (V)"] * 1e5, lw=1.5)
axes[1].set_xlabel(r"$T$ (K)")
axes[1].set_ylabel(r"$R_{xx}$ (Ohm)")
axes[1].set_title("FGT/OFGT — Longitudinal voltage")

fig.tight_layout()
plt.show()
```
